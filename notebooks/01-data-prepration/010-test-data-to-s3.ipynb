## Setup

```python
import boto3
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor

SRC = Path("/mnt/c/Users/HP/Downloads/ISIC_2019_Test_Input/ISIC_2019_Test_Input")
BUCKET = "skin-lesion-data-bucket"
PREFIX = "new-unseen-images/images"

s3 = boto3.client("s3")

```

```python
def upload(p: Path):
    s3.upload_file(str(p), BUCKET, f"{PREFIX}/{p.name}",
                   ExtraArgs={"ContentType": "image/jpeg"})
    return p.name

files = sorted(SRC.glob("*.jpg"))
print(f"{len(files)} files")

with ThreadPoolExecutor(max_workers=16) as pool:
    for i, name in enumerate(pool.map(upload, files), 1):
        if i % 100 == 0:
            print(f"{i}/{len(files)}")

print("done")
```
